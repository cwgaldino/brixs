# Support functions

## Process function

```python
from scipy.ndimage import median_filter, gaussian_filter
def process(scan, dark, slope='auto', calib='auto',
            x_start=None, x_stop=None,
            start=None, stop=None,
            i0_norm=True,
            remove_cosmic_rays=True,
            cosmic_rays_contrast_ratio=8,
            cosmic_rays_nrows2bin=2,
            cosmic_rays_median_filter_window=100,
            cosmic_rays_moving_average_window=20,
            cosmic_rays_minimum_intensity=600,
            cosmic_rays_minimum_average=400,
            dark_smooth=True,
            dark_smooth_median_filter_window=3,
            dark_smooth_sigma=1):
    """
    Process an I21 RIXS scan and return the final spectrum.

    Processing workflow::

        Dark frames
            load
            └─ crop
               └─ optional cosmic-ray removal
                  └─ optional smoothing
                     └─ average
                        └─ d3

        Scan frames
            load
            └─ crop
               └─ optional cosmic-ray removal
                  └─ dark subtraction (using d3)
                     └─ I0 normalization
                        └─ average
                           └─ curvature correction
                              └─ row integration
                                 └─ energy calibration
                                    └─ Spectrum

    Notes
    -----
    * Dark-frame offsets are determined automatically by matching the
      average intensity in the detector region y=1500:1700.
    * If ``i0_norm=True`` but I0 data are unavailable, normalization is
      disabled automatically.
    * Processed spectra and dark images can be cached through
      ``br.finder`` to avoid repeated processing.

    Parameters
    ----------
    scan : int
        Scan number containing the RIXS data.

    dark : int
        Scan number containing the dark images.

    slope : float or 'auto', optional
        Detector curvature correction slope. If ``'auto'``,
        ``settings.SLOPE`` is used.

    calib : float or 'auto', optional
        Energy calibration in eV/pixel. If ``'auto'``,
        ``settings.CALIB`` is used.

    x_start, x_stop : int, optional
        Detector x-range used during processing. If None, the full range
        is used.

    start, stop : int, optional
        First and last frame to process. ``start`` is inclusive and
        ``stop`` is exclusive. If both are None, all frames are used.

    i0_norm : bool, optional
        If True, normalize each image by its I0 value before averaging.

    remove_cosmic_rays : bool, optional
        If True, detect and patch cosmic-ray events in individual frames.

    cosmic_rays_contrast_ratio : float, optional
        Minimum contrast ratio required for outlier detection.

    cosmic_rays_nrows2bin : int, optional
        Number of detector rows combined during cosmic-ray detection.

    cosmic_rays_median_filter_window : int, optional
        Horizontal median-filter window used to estimate local
        background during cosmic-ray detection.

    cosmic_rays_moving_average_window : int, optional
        Moving-average window used for residual normalization during
        cosmic-ray detection.

    cosmic_rays_minimum_intensity : float, optional
        Minimum neighbouring pixel intensity required to validate a
        cosmic-ray candidate.

    cosmic_rays_minimum_average : float, optional
        Minimum neighbouring average intensity required to validate a
        cosmic-ray candidate.

    dark_smooth : bool, optional
        If True, smooth dark-frame images before averaging.

    dark_smooth_median_filter_window : int, optional
        Median-filter window applied to dark-frame images.

    dark_smooth_sigma : float, optional
        Gaussian smoothing sigma applied after median filtering.
        Set to 0 to disable Gaussian smoothing.

    Returns
    -------
    br.Spectrum
        Processed and calibrated RIXS spectrum.

    Cache
    -----
    This function uses ``br.finder`` to cache processed spectra and dark
    images. When the same scan is processed again using identical
    parameters, the cached result is returned instead of rerunning the
    processing pipeline.

    Typical configuration::

        br.finder.folderpath = "/path/to/cache"
        br.finder.verbose = False

    To force reprocessing or disable saving, configure the finder cache
    options accordingly.
    """
    if slope == 'auto': slope = i21.settings.SLOPE
    if calib == 'auto': calib = i21.settings.CALIB
        
    # finder
    br.finder.kwargs = vars()
    del br.finder.kwargs['calib']
    s0 = br.finder.search()
    _search_string = br.finder._search_string

    # if scan was already processed but is still running, rerun the processing
    if s0:
        if bool(s0.metadata['general']['finished']) == False:
            s0 = False

    # processing
    if s0 == False:
        # check finder for dark image
        br.finder.kwargs = vars()
        for _name in ['_search_string', 'scan', 'slope', 'calib']:
            del br.finder.kwargs[_name]
        br.finder.kwargs['darkimage'] = True
        d3 = br.finder.search()
        if d3 == False:
            # dark image loading and cosmic rays removal
            data = i21.readrixs(scan=dark)
            d0  = data['im'].crop(x_start=x_start, x_stop=x_stop)
            ds0 = br.Dummy([_.crop(x_start=x_start, x_stop=x_stop) for _ in data['ims']])
            
            # dark image cosmic rays removal
            ds1   = br.Dummy()
            dpes1 = br.Dummy()
            for _im in ds0:
                if remove_cosmic_rays:
                    _pe  = _im.detect_outliers_via_threshold(threshold=400)
                    try:
                        _im1 = _im.patch2(_pe, n=4)
                    except AssertionError:
                        print(f'ERROR: scan {dark}')
                        _im1 = _im
                    dpes1.append(_pe)  # save cosmics rays in a list for later verification
                else:
                    _im1 = _im 
                ds1.append(_im1)

            # smooth
            ds2 = br.Dummy()
            for _im in ds1:
                if dark_smooth:
                    _img_clean = median_filter(_im.data, size=dark_smooth_median_filter_window)
                    if dark_smooth_sigma > 0:
                        _img_smooth = gaussian_filter(_img_clean, sigma=dark_smooth_sigma)
                    else:
                        _img_smooth = _img_clean
                    ds2.append(br.Image(data=_img_smooth))
                else:
                    ds2.append(_im)
            
            # average all dark frames
            d3 = ds2[0]
            for _im in ds2[1:]:
                d3 += _im
            d3 = d3.set_factor(1/len(ds2))
            
            # finder: save dark image
            br.finder.save(d3)
            # loaded_dark[dark_name] = {'d0': d0, 'ds0': ds0, 'ds1': ds1, 'dpes1': dpes1}

        # initial data
        data = i21.readrixs(scan=scan)

        # check i0
        try:
            i0 = data['im'].metadata['instrument']['i0']
            if i0 is None:
                i0_norm = False
                print(f'scan {scan} does not have i0 data. Defaulting to i0_norm=False.')
        except: 
            i0_norm = False
            print(f'scan {scan} does not have i0 data. Defaulting to i0_norm=False.')
            
        # get im and ims
        if start is None and stop is None:
            im0  = data['im'].crop(x_start=x_start, x_stop=x_stop)
            ims0 = br.Dummy([_.crop(x_start=x_start, x_stop=x_stop) for _ in data['ims']])
        else:
            if start is None:
                start = 0
            if stop is None:
                stop = len(data['ims'])
            ims0 = br.Dummy([_.crop(x_start=x_start, x_stop=x_stop) for _ in data['ims'][start:stop]])
            im0 = ims0[0]
            for i, _im in enumerate(ims0[1:]):
                im0 += _im
            im0.metadata = data['im'].metadata
            if i0_norm:
                im0.metadata['instrument']['i0'] = im0.metadata['instrument']['i0'][start:stop]
            
        # cosmic ray removal
        ims1 = br.Dummy()
        pes1 = br.Dummy()
        for j, _im in enumerate(ims0):
            if remove_cosmic_rays:
                _pe = _im.detect_outliers_via_row_median(contrast_ratio=cosmic_rays_contrast_ratio, slope=slope, nrows2bin=cosmic_rays_nrows2bin,
                                                        median_filter_window=cosmic_rays_median_filter_window, moving_average_window=cosmic_rays_moving_average_window,
                                                        minimum_intensity=cosmic_rays_minimum_intensity, minimum_average=cosmic_rays_minimum_average)
                try:
                    _im1 = _im.patch2(_pe, n=4)
                except AssertionError:
                    print(f'ERROR: scan {scan}:{j}')
                    _im1 = _im
                pes1.append(_pe)  # save cosmics rays in a list for later verification
            else:
                _im1 = _im
            ims1.append(_im1)

        # dark image subtraction
        ims2 = br.Dummy()
        dark_offsets = []
        for _im in ims1:
            imavg = _im.crop(y_start=1500, y_stop=1700).calculate_average()
            davg  = d3.crop(y_start=1500, y_stop=1700).calculate_average()
            dark_offset = imavg-davg
            dark_offsets.append(dark_offset)
            ims2.append(_im - d3.set_offset(dark_offset))

        # plot dark image subtraction verification
        if False:
            for i in range(len(ims2)):
                fig, axes = br.subplots(1, 3, figsize=(28, 12), layout='constrained')
                plt.suptitle(f'scan image {i}')
                ims1[i].integrated_rows_vs_y_centers().plot(ax=axes[0])
                d3.set_offset(dark_offsets[i]).integrated_rows_vs_y_centers().plot(ax=axes[0])    
                # d3.integrated_rows_vs_y_centers().plot(ax=axes[0])    
                ims2[i].integrated_rows_vs_y_centers().plot(ax=axes[1])
                ims2[i].plot(ax=axes[2])
        
        # align images
        # <not needed, only single frame scans>
        
        # average all images
        if i0_norm:
            im2 = ims2[0].set_factor(1/im0.metadata['instrument']['i0'][0])
            for i, _im in enumerate(ims2[1:]):
                im2 += _im.set_factor(1/im0.metadata['instrument']['i0'][i])
            im2 = im2.set_factor(1/len(ims2))
        else:
            im2 = ims2[0]
            for i, _im in enumerate(ims2[1:]):
                im2 += _im
            im2 = im2.set_factor(1/len(ims2))
            

        # slope correction
        im3 = im2.set_vertical_shift_via_polyval(p=[-slope, 0])
        
        # calculate spectrum
        s0 = im3.integrated_rows_vs_y_centers()

        # copy metadata
        s0.metadata = im0.metadata

        # finder: save dark image
        br.finder._search_string = _search_string
        br.finder.save(s0)
    
    # energy calibration
    if calib:
        s1 = s0.set_calib(calib)
    else:
        s1 = s0

    # zero alingment
    # try:
    #     # initial alignment
    #     fit = s1.fit_peak(guess_A=30, guess_c=9.819, guess_w=58/1000, fixed_m=0, limits=(9.471, 10.191))
    #     s1 = s1.set_shift(-fit['popt']['c'])

    #     # refined alignment
    #     if E 
    #     # s1.resolution = round(fit['popt']['fwhm']*1000, 2)
    #     # s1.fit = fit['fit'].set_shift(-fit['popt']['c'])
    #     # s1 = s1.set_shift(-fit['popt']['c'])
    # except:
    #     pass

    return s1
# %%
```

```python
def temp(scan):
    data = i21.read(scan)
    im0 = data['im']
    im1 = data['im'].crop(x_start=500, x_stop=1750, y_start=750, y_stop=1250)
    im2 = im1.binning(3)
    ss = im2.get_columns().floor(limits=(None, 900))
    shift = ss.floor(limits=(None, 900)).calculate_shift(mode='peak')
    s = br.Spectrum(x=im2.x_centers, y=shift)
    polyfit = s.polyfit(1)
    slope = polyfit['popt'][0]
    final = im1.set_vertical_shift_via_polyval([slope, 0]).integrated_rows_vs_y_centers().floor(limits=(None, 900))
    fit = final.fit_peak(fixed_m=0)
    return data['im'].metadata['instrument']['sgmpitch'], fit['popt']['fwhm']
```
